```python
from getpass import getpass
import os

os.environ["GEMINI_API_KEY"] = getpass("Enter your Gemini API key: ")
```

```python
from google import genai
import os

# Initialize Gemini client
client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

model_name = "gemini-3.8-flash"

print("Starting Experiment 03: Iterative Prompt Refinement")
print("=" * 70)


# ==========================================
# ROUND A: MINIMAL INSTRUCTION
# ==========================================

prompt_a = "Summarize the plot of the Shakespearean play Romeo and Juliet."

print("\n[Executing Round A] Minimal Instruction...")

response_a = client.interactions.create(
    model=model_name,
    input=prompt_a
)

output_a = response_a.output_text


# ==========================================
# ROUND B: LENGTH + STYLE CONSTRAINTS
# ==========================================

prompt_b = """
Summarize the plot of the Shakespearean play Romeo and Juliet.

Your output must follow these explicit constraints:

1. Length: Exactly two sentences long.

2. Style: Write in a dramatic, poetic, and elevated theatrical tone.
"""

print("[Executing Round B] Length and Style Constraints Added...")

response_b = client.interactions.create(
    model=model_name,
    input=prompt_b
)

output_b = response_b.output_text


# ==========================================
# ROUND C: KEY CONTENT ELEMENTS
# ==========================================

prompt_c = """
Summarize the plot of the Shakespearean play Romeo and Juliet.

Your output must follow these explicit constraints:

1. Length: Exactly two sentences long.

2. Style: Write in a dramatic, poetic, and elevated theatrical tone.

3. Content: You must explicitly mention the setting (Verona)
and highlight the core theme (how ancient family hatred destroys young love).
"""

print("[Executing Round C] Setting and Theme Constraints Added...")

response_c = client.interactions.create(
    model=model_name,
    input=prompt_c
)

output_c = response_c.output_text


# ==========================================
# EXPERIMENT DOCUMENTATION
# ==========================================

print("\n" + "#" * 70)
print("EXPERIMENT DOCUMENTATION REPORT")
print("#" * 70)


# ROUND A
print("\nITERATION A: Naive / Minimal Prompt")
print("-" * 50)

print("Prompt:")
print(prompt_a)

print("\nOutput:")
print(output_a)

print("\nImpact Analysis:")
print(
    "The model provides a general summary. "
    "Without constraints, the response length and style can vary."
)


# ROUND B
print("\n" + "-" * 70)
print("ITERATION B: Formatting & Length Enforcement")
print("-" * 70)

print("Prompt:")
print(prompt_b)

print("\nOutput:")
print(output_b)

print("\nImpact Analysis:")
print(
    "Adding the two-sentence constraint controls the length, "
    "while the dramatic and poetic instruction changes the writing style."
)


# ROUND C
print("\n" + "-" * 70)
print("ITERATION C: Fully Optimized Contextual Prompt")
print("-" * 70)

print("Prompt:")
print(prompt_c)

print("\nOutput:")
print(output_c)

print("\nImpact Analysis:")
print(
    "Adding Verona and the theme of family hatred provides additional "
    "context and makes the response more specific to the intended task."
)
```
